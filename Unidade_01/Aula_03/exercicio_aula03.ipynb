## 💻 Desenvolvimento Prático

### Bloco 1 — Configuração e Execução Inicial

```python
# Cálculo de faturamento e ticket médio
import pandas as pd

vendas = pd.DataFrame({
    "pedido": [1, 1, 2, 3],
    "quantidade": [2, 3, 1, 4],
    "preco_unitario": [10.0, 5.0, 50.0, 12.5]
})
vendas["receita"] = vendas["quantidade"] * vendas["preco_unitario"]
total_faturamento = vendas["receita"].sum()
print(f"Faturamento total: R$ {total_faturamento:.2f}")
```

#### 🔍 Análise e Interpretação do Bloco 1
> O faturamento considera vendas com quantidades e preços válidos. Cancelamentos e ajustes de estoque devem ser devidamente isolados para evitar dupla contagem ou faturamento artificial.
