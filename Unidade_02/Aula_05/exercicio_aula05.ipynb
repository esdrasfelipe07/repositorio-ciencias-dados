## 💻 Desenvolvimento Prático

### Bloco 1 — Configuração e Execução Inicial

```python
import numpy as np
import scipy.stats as stats

pedidos = np.array([10, 12, 15, 18, 20, 22, 25, 30, 250])
print(f"Média: {np.mean(pedidos):.2f}")
print(f"Mediana: {np.median(pedidos):.2f}")
print(f"Assimetria (skew): {stats.skew(pedidos):.2f}")
print(f"Curtose: {stats.kurtosis(pedidos):.2f}")
```

#### 🔍 Análise e Interpretação do Bloco 1
> A presença do valor 250 (outlier) puxa a média para cima enquanto a mediana permanece estável em 20. O skew positivo indica distribuição com cauda longa à direita.
