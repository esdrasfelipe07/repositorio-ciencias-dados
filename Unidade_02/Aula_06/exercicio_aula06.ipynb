## 💻 Desenvolvimento Prático

### Bloco 1 — Configuração e Execução Inicial

```python
from sklearn.preprocessing import StandardScaler, RobustScaler
import numpy as np

alugueis = np.array([[1200], [1500], [1800], [2200], [15000]])
std_s = StandardScaler().fit_transform(alugueis)
rob_s = RobustScaler().fit_transform(alugueis)
print("StandardScaler:", std_s.ravel())
print("RobustScaler:", rob_s.ravel())
```

#### 🔍 Análise e Interpretação do Bloco 1
> O RobustScaler utiliza a mediana e o IQR, sendo muito mais resistente ao imóvel de R$ 15.000 do que o StandardScaler, que tem sua média e variância distorcidas pelo extremo.
