## 💻 Desenvolvimento Prático

### Bloco 1 — Configuração e Execução Inicial

```python
import numpy as np
from scipy import stats

grupo_a = [21.0, 21.0, 22.8, 21.4, 18.7]
grupo_b = [14.3, 16.4, 17.3, 15.2, 10.4]
t_stat, p_val = stats.ttest_ind(grupo_a, grupo_b)
print(f"Estatística t: {t_stat:.3f}, p-valor: {p_val:.4f}")
```

#### 🔍 Análise e Interpretação do Bloco 1
> Com o p-valor inferior a 0.05, rejeitamos a hipótese nula H0 de médias iguais, confirmando diferença estatisticamente significante entre os grupos amostrados.
