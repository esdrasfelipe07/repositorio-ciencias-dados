## 💻 Desenvolvimento Prático

### Bloco 1 — Configuração e Execução Inicial

```python
from scipy import stats

grupo_tratamento = [5, 6, 7, 8, 9, 120] # presença de outlier extremo
grupo_controle = [1, 2, 3, 4, 4, 5]
u_stat, p_val = stats.mannwhitneyu(grupo_tratamento, grupo_controle, alternative="two-sided")
print(f"Mann-Whitney U: {u_stat}, p-valor: {p_val:.4f}")
```

#### 🔍 Análise e Interpretação do Bloco 1
> O teste não paramétrico de postos preserva a sensibilidade estatística mesmo com o outlier severo de 120 dias, onde a média e o teste t falhariam gravemente.
