## 💻 Desenvolvimento Prático

### Bloco 1 — Configuração e Execução Inicial

```python
from scipy import stats

# Teste t de Welch (variâncias desiguais)
amostra_1 = [25, 28, 30, 32, 29, 31]
amostra_2 = [18, 20, 19, 22, 21, 20]
res = stats.ttest_ind(amostra_1, amostra_2, equal_var=False)
print(f"Welch t-test: t = {res.statistic:.3f}, p = {res.pvalue:.4e}")
```

#### 🔍 Análise e Interpretação do Bloco 1
> O teste de Welch relaxa a premissa de variâncias homogêneas, fornecendo um teste robusto para comparação de médias entre dois grupos independentes.
